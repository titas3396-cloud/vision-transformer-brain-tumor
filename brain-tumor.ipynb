```python
import warnings
warnings.filterwarnings("ignore")

# import system libs
import os
import itertools
from PIL import Image

# import data handling tools
import numpy as np
import pandas as pd
import seaborn as sns
sns.set_style('darkgrid')
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# import Deep learning Libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print('modules loaded')


# ============================================================
# Data Preprocessing
# ============================================================

# Read training data and store paths with labels in a dataframe
train_data_dir = '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Training'
filepaths = []
labels = []

folds = os.listdir(train_data_dir)
for fold in folds:
    foldpath = os.path.join(train_data_dir, fold)
    filelist = os.listdir(foldpath)
    for file in filelist:
        fpath = os.path.join(foldpath, file)
        filepaths.append(fpath)
        labels.append(fold)

# Concatenate data paths with labels into one dataframe
Fseries = pd.Series(filepaths, name='filepaths')
Lseries = pd.Series(labels, name='labels')
train_df = pd.concat([Fseries, Lseries], axis=1)


# Read testing data and store paths with labels in a dataframe
test_data_dir = '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Testing'
filepaths = []
labels = []

folds = os.listdir(test_data_dir)
for fold in folds:
    foldpath = os.path.join(test_data_dir, fold)
    filelist = os.listdir(foldpath)
    for file in filelist:
        fpath = os.path.join(foldpath, file)
        filepaths.append(fpath)
        labels.append(fold)

# Concatenate data paths with labels into one dataframe
Fseries = pd.Series(filepaths, name='filepaths')
Lseries = pd.Series(labels, name='labels')
ts_df = pd.concat([Fseries, Lseries], axis=1)


# Split test dataframe into valid and test
valid_df, test_df = train_test_split(ts_df, train_size=0.5, shuffle=True, random_state=123)


# ============================================================
# Image Data Generators
# ============================================================

batch_size = 16
img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)

tr_gen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.06,
    height_shift_range=0.06,
    zoom_range=0.12,
    horizontal_flip=True,
    fill_mode='nearest')
ts_gen = ImageDataGenerator()

train_gen = tr_gen.flow_from_dataframe(train_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

valid_gen = ts_gen.flow_from_dataframe(valid_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

test_gen = ts_gen.flow_from_dataframe(test_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                      color_mode='rgb', shuffle=False, batch_size=batch_size)


# ============================================================
# Show sample from train data
# ============================================================

g_dict = train_gen.class_indices       # defines dictionary {'class': index}
classes = list(g_dict.keys())          # list of class names
images, labels = next(train_gen)       # get a batch of samples from the generator

plt.figure(figsize=(20, 20))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    image = images[i] / 255            # scale data to range (0 - 1)
    plt.imshow(image)
    index = np.argmax(labels[i])       # get image index
    class_name = classes[index]        # get class of image
    plt.title(class_name, color='blue', fontsize=12)
    plt.axis('off')
plt.tight_layout()
plt.show()


# ============================================================
# Model Structure
#   Hierarchical Vision Transformer (HViT)
#     - 4 hierarchical stages: 56 -> 28 -> 14 -> 7  (dims 64/128/256/512)
#     - Spatial-Reduction multi-head self-attention (PVT-style) per block
#     - Dual-Branch Spatial Attention module inside every block
#       (avg-pooled channel descriptor branch + max-pooled channel branch,
#        merged into a learned spatial gate)
#     - Global average pooling + linear classification head
# ============================================================

img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)
class_count = len(list(train_gen.class_indices.keys()))


DENSE_INIT = keras.initializers.TruncatedNormal(stddev=0.02)


class DropPath(layers.Layer):
    """Stochastic depth: randomly drops the residual branch per sample."""
    def __init__(self, rate=0.0, **kwargs):
        super().__init__(**kwargs)
        self.rate = rate

    def call(self, x, training=False):
        if (not training) or self.rate == 0.0:
            return x
        keep = 1.0 - self.rate
        shape = [tf.shape(x)[0]] + [1] * (len(x.shape) - 1)
        mask = tf.floor(keep + tf.random.uniform(shape, dtype=x.dtype))
        return x / keep * mask


class OverlapPatchEmbed(layers.Layer):
    """Overlapping patch embedding (downsampling) via a strided conv."""
    def __init__(self, embed_dim, patch_size, stride, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.proj = layers.Conv2D(embed_dim, patch_size, strides=stride, padding='same')
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x):
        x = self.proj(x)                          # B, H, W, C
        H = tf.shape(x)[1]
        W = tf.shape(x)[2]
        x = tf.reshape(x, [tf.shape(x)[0], H * W, self.embed_dim])
        x = self.norm(x)
        return x, H, W


class Attention(layers.Layer):
    """Multi-head self-attention with spatial reduction on keys/values."""
    def __init__(self, dim, num_heads, sr_ratio=1, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.sr_ratio = sr_ratio

        self.q = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        self.kv = layers.Dense(dim * 2, kernel_initializer=DENSE_INIT)
        self.proj = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        if sr_ratio > 1:
            self.sr = layers.Conv2D(dim, sr_ratio, strides=sr_ratio, padding='same')
            self.sr_norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, H, W, return_attn=False):
        B = tf.shape(x)[0]
        N = tf.shape(x)[1]

        q = self.q(x)
        q = tf.reshape(q, [B, N, self.num_heads, self.head_dim])
        q = tf.transpose(q, [0, 2, 1, 3])                        # B, heads, N, hd

        if self.sr_ratio > 1:
            x_ = tf.reshape(x, [B, H, W, self.dim])
            x_ = self.sr(x_)                                     # B, H', W', C
            x_ = tf.reshape(x_, [B, -1, self.dim])
            x_ = self.sr_norm(x_)
            kv = self.kv(x_)
        else:
            kv = self.kv(x)

        Nkv = tf.shape(kv)[1]
        kv = tf.reshape(kv, [B, Nkv, 2, self.num_heads, self.head_dim])
        kv = tf.transpose(kv, [2, 0, 3, 1, 4])                   # 2, B, heads, Nkv, hd
        k, v = kv[0], kv[1]

        attn = tf.matmul(q, k, transpose_b=True) * self.scale    # B, heads, N, Nkv
        attn = tf.nn.softmax(attn, axis=-1)

        out = tf.matmul(attn, v)                                 # B, heads, N, hd
        out = tf.transpose(out, [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.dim])
        out = self.proj(out)

        if return_attn:
            return out, attn
        return out


class DualBranchSpatialAttention(layers.Layer):
    """Dual-branch spatial attention: an average-pooled channel branch and a
    max-pooled channel branch are each convolved and merged into a spatial gate."""
    def __init__(self, dim, kernel_size=7, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.branch_avg = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)
        self.branch_max = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)

    def call(self, x, H, W):
        B = tf.shape(x)[0]
        f = tf.reshape(x, [B, H, W, self.dim])
        avg = tf.reduce_mean(f, axis=-1, keepdims=True)          # channel-avg descriptor
        mx = tf.reduce_max(f, axis=-1, keepdims=True)            # channel-max descriptor
        gate = tf.sigmoid(self.branch_avg(avg) + self.branch_max(mx))   # B, H, W, 1
        f = f * gate
        return tf.reshape(f, [B, H * W, self.dim])


class MLP(layers.Layer):
    def __init__(self, dim, hidden_dim, **kwargs):
        super().__init__(**kwargs)
        self.fc1 = layers.Dense(hidden_dim, kernel_initializer=DENSE_INIT)
        self.fc2 = layers.Dense(dim, kernel_initializer=DENSE_INIT)

    def call(self, x):
        x = self.fc1(x)
        x = tf.nn.gelu(x)
        x = self.fc2(x)
        return x


class Block(layers.Layer):
    """Transformer block = SR-MHSA + Dual-Branch Spatial Attention + MLP (all residual)."""
    def __init__(self, dim, num_heads, mlp_ratio=4, sr_ratio=1, drop_path=0.0, **kwargs):
        super().__init__(**kwargs)
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = Attention(dim, num_heads, sr_ratio)
        self.norm_sa = layers.LayerNormalization(epsilon=1e-6)
        self.dbsa = DualBranchSpatialAttention(dim)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = MLP(dim, dim * mlp_ratio)
        self.drop_path = DropPath(drop_path)

    def call(self, x, H, W, return_attn=False, training=False):
        if return_attn:
            y, attn = self.attn(self.norm1(x), H, W, return_attn=True)
        else:
            y = self.attn(self.norm1(x), H, W)
        x = x + self.drop_path(y, training=training)
        x = x + self.drop_path(self.dbsa(self.norm_sa(x), H, W), training=training)
        x = x + self.drop_path(self.mlp(self.norm2(x)), training=training)
        if return_attn:
            return x, attn
        return x


class Stage(layers.Layer):
    def __init__(self, embed_dim, depth, num_heads, mlp_ratio, sr_ratio, patch_size, stride,
                 drop_path_rates=None, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.patch_embed = OverlapPatchEmbed(embed_dim, patch_size, stride)
        if drop_path_rates is None:
            drop_path_rates = [0.0] * depth
        self.blocks = [Block(embed_dim, num_heads, mlp_ratio, sr_ratio, drop_path=drop_path_rates[i])
                       for i in range(depth)]
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, return_attn=False, training=False):
        x, H, W = self.patch_embed(x)
        attns = []
        for blk in self.blocks:
            if return_attn:
                x, a = blk(x, H, W, return_attn=True, training=training)
                attns.append(a)
            else:
                x = blk(x, H, W, training=training)
        x = self.norm(x)
        x_img = tf.reshape(x, [tf.shape(x)[0], H, W, self.embed_dim])
        return x_img, attns, H, W


class HViT(Model):
    def __init__(self, num_classes,
                 embed_dims=(64, 128, 256, 512),
                 depths=(2, 2, 2, 2),
                 num_heads=(1, 2, 4, 8),
                 sr_ratios=(8, 4, 2, 1),
                 mlp_ratio=4, drop_path=0.1, **kwargs):
        super().__init__(**kwargs)
        self.rescale = layers.Rescaling(1. / 255)
        # linearly increasing stochastic-depth rate across all blocks
        dpr = list(np.linspace(0.0, drop_path, sum(depths)))
        s = [0] + list(np.cumsum(depths))
        self.stage1 = Stage(embed_dims[0], depths[0], num_heads[0], mlp_ratio, sr_ratios[0], 7, 4, dpr[s[0]:s[1]])
        self.stage2 = Stage(embed_dims[1], depths[1], num_heads[1], mlp_ratio, sr_ratios[1], 3, 2, dpr[s[1]:s[2]])
        self.stage3 = Stage(embed_dims[2], depths[2], num_heads[2], mlp_ratio, sr_ratios[2], 3, 2, dpr[s[2]:s[3]])
        self.stage4 = Stage(embed_dims[3], depths[3], num_heads[3], mlp_ratio, sr_ratios[3], 3, 2, dpr[s[3]:s[4]])
        self.head = layers.Dense(num_classes, activation='softmax', kernel_initializer=DENSE_INIT)

    def call(self, x, return_attn=False, training=False):
        x = self.rescale(x)
        attns = []
        x, a, H, W = self.stage1(x, return_attn=return_attn, training=training); attns += [(z, H, W) for z in a]
        x, a, H, W = self.stage2(x, return_attn=return_attn, training=training); attns += [(z, H, W) for z in a]
        x, a, H, W = self.stage3(x, return_attn=return_attn, training=training); attns += [(z, H, W) for z in a]
        x, a, H, W = self.stage4(x, return_attn=return_attn, training=training); attns += [(z, H, W) for z in a]
        x = tf.reduce_mean(x, axis=[1, 2])          # global average pooling
        logits = self.head(x)
        if return_attn:
            return logits, attns
        return logits


def build_model(num_classes):
    m = HViT(num_classes)
    _ = m(tf.zeros((1, img_size[0], img_size[1], channels)))   # build the graph
    return m


model = build_model(class_count)
model.summary()


# ============================================================
# Training
# ============================================================

epochs = 60
steps_per_epoch = int(np.ceil(train_gen.samples / batch_size))
total_steps = steps_per_epoch * epochs
warmup_steps = steps_per_epoch * 5          # 5-epoch linear warmup (critical for ViT stability)

# Cosine schedule WITH warmup: LR ramps 1e-6 -> 3e-4 then cosine-decays.
try:
    lr_schedule = keras.optimizers.schedules.CosineDecay(
        initial_learning_rate=1e-6,
        warmup_target=3e-4,
        warmup_steps=warmup_steps,
        decay_steps=max(total_steps - warmup_steps, 1),
        alpha=0.02)
except TypeError:
    # Older Keras without warmup support
    lr_schedule = keras.optimizers.schedules.CosineDecay(
        initial_learning_rate=3e-4, decay_steps=total_steps, alpha=0.02)

try:
    optimizer = keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=5e-2)
except Exception:
    optimizer = Adam(learning_rate=lr_schedule)

model.compile(optimizer=optimizer,
              loss=keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
              metrics=['accuracy'])

# Balance the classes (no_tumor is under-represented) so the model can't win by
# collapsing to the majority class.
from sklearn.utils.class_weight import compute_class_weight
_cls = np.unique(train_gen.classes)
_cw = compute_class_weight('balanced', classes=_cls, y=train_gen.classes)
class_weights = {int(c): float(w) for c, w in zip(_cls, _cw)}
print("class weights:", class_weights)

history = model.fit(x=train_gen, epochs=epochs, verbose=1, validation_data=valid_gen,
                    class_weight=class_weights, shuffle=False)


# ============================================================
# Plot Training History
# ============================================================

tr_acc = history.history['accuracy']
tr_loss = history.history['loss']
val_acc = history.history['val_accuracy']
val_loss = history.history['val_loss']
index_loss = np.argmin(val_loss)
val_lowest = val_loss[index_loss]
index_acc = np.argmax(val_acc)
acc_highest = val_acc[index_acc]

Epochs = [i + 1 for i in range(len(tr_acc))]
loss_label = f'best epoch= {str(index_loss + 1)}'
acc_label = f'best epoch= {str(index_acc + 1)}'

plt.figure(figsize=(20, 8))
plt.style.use('fivethirtyeight')

plt.subplot(1, 2, 1)
plt.plot(Epochs, tr_loss, 'r', label='Training loss')
plt.plot(Epochs, val_loss, 'g', label='Validation loss')
plt.scatter(index_loss + 1, val_lowest, s=150, c='blue', label=loss_label)
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(Epochs, tr_acc, 'r', label='Training Accuracy')
plt.plot(Epochs, val_acc, 'g', label='Validation Accuracy')
plt.scatter(index_acc + 1, acc_highest, s=150, c='blue', label=acc_label)
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# Evaluate
# ============================================================

train_score = model.evaluate(train_gen, verbose=1)
valid_score = model.evaluate(valid_gen, verbose=1)
test_score = model.evaluate(test_gen, verbose=1)

print("Train Loss: ", train_score[0])
print("Train Accuracy: ", train_score[1])
print('-' * 20)
print("Validation Loss: ", valid_score[0])
print("Validation Accuracy: ", valid_score[1])
print('-' * 20)
print("Test Loss: ", test_score[0])
print("Test Accuracy: ", test_score[1])


# ============================================================
# Predictions and Confusion Matrix
# ============================================================

preds = model.predict(test_gen)
y_pred = np.argmax(preds, axis=1)

g_dict = test_gen.class_indices
classes = list(g_dict.keys())

# Confusion matrix
cm = confusion_matrix(test_gen.classes, y_pred)

plt.figure(figsize=(10, 10))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()

tick_marks = np.arange(len(classes))
plt.xticks(tick_marks, classes, rotation=45)
plt.yticks(tick_marks, classes)

thresh = cm.max() / 2.
for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
    plt.text(j, i, cm[i, j], horizontalalignment='center', color='white' if cm[i, j] > thresh else 'black')

plt.tight_layout()
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

# Classification report
print(classification_report(test_gen.classes, y_pred, target_names=classes))


# ============================================================
# Save and Reload Model  (subclassed model -> save/load weights)
# ============================================================

weights_path = '/kaggle/working/HViT_BrainTumor.weights.h5'
model.save_weights(weights_path)

loaded_model = build_model(class_count)
loaded_model.load_weights(weights_path)
loaded_model.compile(optimizer=Adam(learning_rate=1e-4),
                     loss='categorical_crossentropy', metrics=['accuracy'])


# ============================================================
# Single Image Prediction
# ============================================================

# map generator class indices -> names (correct, ordered mapping)
idx_to_class = {v: k for k, v in train_gen.class_indices.items()}

# take a test image from the test dataframe (robust; no external path needed)
image_path = test_df['filepaths'].iloc[0]
image = Image.open(image_path).convert('RGB')

img = image.resize((224, 224))
img_array = tf.keras.preprocessing.image.img_to_array(img)
img_array = tf.expand_dims(img_array, 0)          # model rescales internally

predictions = loaded_model.predict(img_array)
pred_idx = int(np.argmax(predictions))
print(f"Predicted: {idx_to_class[pred_idx]}  (true: {test_df['labels'].iloc[0]})")


# ============================================================
# Explainability: Layer-wise Relevance Propagation (LRP)
#   Transformer-adapted class-specific relevance (Chefer-style).
#   For each attention layer, relevance of a query (spatial) token i is
#       R_i = sum_j  ReLU( dS/dA_ij  *  A_ij )   averaged over heads,
#   where S is the target-class score and A the softmax attention. Each stage's
#   relevance is reshaped to its grid, upsampled to 224x224, and averaged
#   across all layers/stages to form the final relevance heatmap.
# ============================================================

def lrp_relevance(m, x, target_class=None):
    x = tf.convert_to_tensor(x, dtype=tf.float32)
    with tf.GradientTape() as tape:
        logits, attns = m(x, return_attn=True, training=False)
        if target_class is None:
            target_class = int(tf.argmax(logits[0]))
        score = logits[:, target_class]

    attn_tensors = [a for (a, H, W) in attns]
    grads = tape.gradient(score, attn_tensors)

    maps = []
    for (a, H, W), g in zip(attns, grads):
        if g is None:
            continue
        r = tf.nn.relu(g * a)                 # B, heads, Nq, Nkv
        r = tf.reduce_sum(r, axis=-1)         # B, heads, Nq   (contribution per query token)
        r = tf.reduce_mean(r, axis=1)         # B, Nq
        r = tf.reshape(r[0], [H, W])          # H, W  (batch = 1)
        r = r / (tf.reduce_max(r) + 1e-8)
        r = tf.image.resize(r[None, ..., None], (224, 224))[0, ..., 0]
        maps.append(r)

    heat = tf.reduce_mean(tf.stack(maps, axis=0), axis=0)
    heat = heat / (tf.reduce_max(heat) + 1e-8)
    return heat.numpy(), target_class


# Visualize LRP heatmaps on a few test images
n_show = 6
plt.figure(figsize=(18, 6))
for k in range(n_show):
    p = test_df['filepaths'].iloc[k]
    true_label = test_df['labels'].iloc[k]
    im = Image.open(p).convert('RGB').resize((224, 224))
    arr = tf.keras.preprocessing.image.img_to_array(im)
    arr_b = tf.expand_dims(arr, 0)

    heat, cls = lrp_relevance(loaded_model, arr_b)
    pred_label = idx_to_class[cls]

    plt.subplot(2, n_show, k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.title(f"{true_label}", fontsize=10)
    plt.axis('off')

    plt.subplot(2, n_show, n_show + k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.imshow(heat, cmap='jet', alpha=0.45)
    plt.title(f"LRP -> {pred_label}", fontsize=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import warnings
warnings.filterwarnings("ignore")

# import system libs
import os
import itertools
from PIL import Image

# import data handling tools
import numpy as np
import pandas as pd
import seaborn as sns
sns.set_style('darkgrid')
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# import Deep learning Libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print('modules loaded')


# ============================================================
# Data Preprocessing
# ============================================================

# NOTE ON THE SPLIT:
# The official Training/ and Testing/ folders of this dataset are drawn from
# DIFFERENT distributions, so training on Training/ and evaluating on Testing/
# gives a large train->test gap (models learn Training/ but fail on Testing/).
# Published 98-99% results on this dataset use k-fold CV or a re-split of the
# POOLED data. We therefore pool both folders and make a stratified
# train/val/test split so all three sets share the same distribution.

data_dirs = [
    '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Training',
    '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Testing',
]

filepaths = []
labels = []
for data_dir in data_dirs:
    for fold in os.listdir(data_dir):
        foldpath = os.path.join(data_dir, fold)
        if not os.path.isdir(foldpath):
            continue
        for file in os.listdir(foldpath):
            filepaths.append(os.path.join(foldpath, file))
            labels.append(fold)

# Concatenate all data paths with labels into one dataframe
Fseries = pd.Series(filepaths, name='filepaths')
Lseries = pd.Series(labels, name='labels')
all_df = pd.concat([Fseries, Lseries], axis=1)
print("total images:", len(all_df))
print(all_df['labels'].value_counts())

# Stratified split: 70% train, 15% validation, 15% test
train_df, temp_df = train_test_split(all_df, train_size=0.70, shuffle=True,
                                     stratify=all_df['labels'], random_state=42)
valid_df, test_df = train_test_split(temp_df, train_size=0.50, shuffle=True,
                                     stratify=temp_df['labels'], random_state=42)
print(f"train: {len(train_df)}  valid: {len(valid_df)}  test: {len(test_df)}")


# ============================================================
# Image Data Generators
# ============================================================

batch_size = 16
img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)

tr_gen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.06,
    height_shift_range=0.06,
    zoom_range=0.12,
    horizontal_flip=True,
    fill_mode='nearest')
ts_gen = ImageDataGenerator()

train_gen = tr_gen.flow_from_dataframe(train_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

valid_gen = ts_gen.flow_from_dataframe(valid_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

test_gen = ts_gen.flow_from_dataframe(test_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                      color_mode='rgb', shuffle=False, batch_size=batch_size)


# ============================================================
# Show sample from train data
# ============================================================

g_dict = train_gen.class_indices       # defines dictionary {'class': index}
classes = list(g_dict.keys())          # list of class names
images, labels = next(train_gen)       # get a batch of samples from the generator

plt.figure(figsize=(20, 20))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    image = images[i] / 255            # scale data to range (0 - 1)
    plt.imshow(image)
    index = np.argmax(labels[i])       # get image index
    class_name = classes[index]        # get class of image
    plt.title(class_name, color='blue', fontsize=12)
    plt.axis('off')
plt.tight_layout()
plt.show()


# ============================================================
# Model Structure
#   Pretrained-backbone Hierarchical Vision Transformer (HViT)
#     - EfficientNetV2-S (ImageNet-pretrained) hierarchical feature extractor
#     - 1x1 projection + learnable positional embedding on the 7x7 feature grid
#     - Transformer blocks (multi-head self-attention) operating on the grid,
#       each with a Dual-Branch Spatial Attention module
#       (avg-pooled channel branch + max-pooled channel branch -> spatial gate)
#     - Global average pooling + linear classification head
#     - Explainability via transformer-adapted Layer-wise Relevance Propagation
#   Requires internet ON in the Kaggle notebook for the ImageNet weights download.
# ============================================================

from tensorflow.keras.applications import EfficientNetV2S

img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)
class_count = len(list(train_gen.class_indices.keys()))


DENSE_INIT = keras.initializers.TruncatedNormal(stddev=0.02)


class DropPath(layers.Layer):
    """Stochastic depth: randomly drops the residual branch per sample."""
    def __init__(self, rate=0.0, **kwargs):
        super().__init__(**kwargs)
        self.rate = rate

    def call(self, x, training=False):
        if (not training) or self.rate == 0.0:
            return x
        keep = 1.0 - self.rate
        shape = [tf.shape(x)[0]] + [1] * (len(x.shape) - 1)
        mask = tf.floor(keep + tf.random.uniform(shape, dtype=x.dtype))
        return x / keep * mask


class Attention(layers.Layer):
    """Multi-head self-attention with spatial reduction on keys/values."""
    def __init__(self, dim, num_heads, sr_ratio=1, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.sr_ratio = sr_ratio

        self.q = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        self.kv = layers.Dense(dim * 2, kernel_initializer=DENSE_INIT)
        self.proj = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        if sr_ratio > 1:
            self.sr = layers.Conv2D(dim, sr_ratio, strides=sr_ratio, padding='same')
            self.sr_norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, H, W, return_attn=False):
        B = tf.shape(x)[0]
        N = tf.shape(x)[1]

        q = self.q(x)
        q = tf.reshape(q, [B, N, self.num_heads, self.head_dim])
        q = tf.transpose(q, [0, 2, 1, 3])                        # B, heads, N, hd

        if self.sr_ratio > 1:
            x_ = tf.reshape(x, [B, H, W, self.dim])
            x_ = self.sr(x_)                                     # B, H', W', C
            x_ = tf.reshape(x_, [B, -1, self.dim])
            x_ = self.sr_norm(x_)
            kv = self.kv(x_)
        else:
            kv = self.kv(x)

        Nkv = tf.shape(kv)[1]
        kv = tf.reshape(kv, [B, Nkv, 2, self.num_heads, self.head_dim])
        kv = tf.transpose(kv, [2, 0, 3, 1, 4])                   # 2, B, heads, Nkv, hd
        k, v = kv[0], kv[1]

        attn = tf.matmul(q, k, transpose_b=True) * self.scale    # B, heads, N, Nkv
        attn = tf.nn.softmax(attn, axis=-1)

        out = tf.matmul(attn, v)                                 # B, heads, N, hd
        out = tf.transpose(out, [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.dim])
        out = self.proj(out)

        if return_attn:
            return out, attn
        return out


class DualBranchSpatialAttention(layers.Layer):
    """Dual-branch spatial attention: an average-pooled channel branch and a
    max-pooled channel branch are each convolved and merged into a spatial gate."""
    def __init__(self, dim, kernel_size=7, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.branch_avg = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)
        self.branch_max = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)

    def call(self, x, H, W):
        B = tf.shape(x)[0]
        f = tf.reshape(x, [B, H, W, self.dim])
        avg = tf.reduce_mean(f, axis=-1, keepdims=True)          # channel-avg descriptor
        mx = tf.reduce_max(f, axis=-1, keepdims=True)            # channel-max descriptor
        gate = tf.sigmoid(self.branch_avg(avg) + self.branch_max(mx))   # B, H, W, 1
        f = f * gate
        return tf.reshape(f, [B, H * W, self.dim])


class MLP(layers.Layer):
    def __init__(self, dim, hidden_dim, **kwargs):
        super().__init__(**kwargs)
        self.fc1 = layers.Dense(hidden_dim, kernel_initializer=DENSE_INIT)
        self.fc2 = layers.Dense(dim, kernel_initializer=DENSE_INIT)

    def call(self, x):
        x = self.fc1(x)
        x = tf.nn.gelu(x)
        x = self.fc2(x)
        return x


class Block(layers.Layer):
    """Transformer block = SR-MHSA + Dual-Branch Spatial Attention + MLP (all residual)."""
    def __init__(self, dim, num_heads, mlp_ratio=4, sr_ratio=1, drop_path=0.0, **kwargs):
        super().__init__(**kwargs)
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = Attention(dim, num_heads, sr_ratio)
        self.norm_sa = layers.LayerNormalization(epsilon=1e-6)
        self.dbsa = DualBranchSpatialAttention(dim)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = MLP(dim, dim * mlp_ratio)
        self.drop_path = DropPath(drop_path)

    def call(self, x, H, W, return_attn=False, training=False):
        if return_attn:
            y, attn = self.attn(self.norm1(x), H, W, return_attn=True)
        else:
            y = self.attn(self.norm1(x), H, W)
        x = x + self.drop_path(y, training=training)
        x = x + self.drop_path(self.dbsa(self.norm_sa(x), H, W), training=training)
        x = x + self.drop_path(self.mlp(self.norm2(x)), training=training)
        if return_attn:
            return x, attn
        return x


class HViT(Model):
    """Pretrained EfficientNetV2-S backbone -> transformer neck with Dual-Branch
    Spatial Attention -> classification head. Exposes attention maps for LRP."""
    def __init__(self, num_classes, dim=256, depth=4, num_heads=8,
                 mlp_ratio=4, drop_path=0.1, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        # EfficientNetV2-S includes its own preprocessing, so it accepts raw 0-255 input.
        self.backbone = EfficientNetV2S(include_top=False, weights='imagenet',
                                        input_shape=(224, 224, 3), include_preprocessing=True)
        self.backbone.trainable = False                      # frozen transfer learning
        self.proj = layers.Conv2D(dim, 1, kernel_initializer=DENSE_INIT)   # 1x1: 1280 -> dim
        self.pos = self.add_weight(name='pos_embed', shape=(1, 7 * 7, dim),
                                   initializer=DENSE_INIT, trainable=True)
        dpr = list(np.linspace(0.0, drop_path, depth))
        self.blocks = [Block(dim, num_heads, mlp_ratio, sr_ratio=1, drop_path=dpr[i])
                       for i in range(depth)]
        self.norm = layers.LayerNormalization(epsilon=1e-6)
        self.drop = layers.Dropout(0.3)
        self.head = layers.Dense(num_classes, activation='softmax', kernel_initializer=DENSE_INIT)

    def call(self, x, return_attn=False, training=False):
        f = self.backbone(x, training=False)                 # B, 7, 7, 1280 (BN kept frozen)
        f = self.proj(f)                                     # B, 7, 7, dim
        H = tf.shape(f)[1]
        W = tf.shape(f)[2]
        x = tf.reshape(f, [tf.shape(f)[0], H * W, self.dim])
        x = x + self.pos
        attns = []
        for blk in self.blocks:
            if return_attn:
                x, a = blk(x, H, W, return_attn=True, training=training)
                attns.append((a, H, W))
            else:
                x = blk(x, H, W, training=training)
        x = self.norm(x)
        x = tf.reduce_mean(x, axis=1)                        # global average pooling over tokens
        x = self.drop(x, training=training)
        logits = self.head(x)
        if return_attn:
            return logits, attns
        return logits

    def unfreeze_backbone(self, from_layer=0.7):
        """Optionally unfreeze the top fraction of backbone layers for fine-tuning."""
        self.backbone.trainable = True
        cut = int(len(self.backbone.layers) * from_layer)
        for i, layer in enumerate(self.backbone.layers):
            layer.trainable = (i >= cut) and (not isinstance(layer, layers.BatchNormalization))


def build_model(num_classes):
    m = HViT(num_classes)
    _ = m(tf.zeros((1, img_size[0], img_size[1], channels)))   # build the graph
    return m


model = build_model(class_count)
model.summary()


# ============================================================
# Training
# ============================================================

# Balance the classes (no_tumor is under-represented).
from sklearn.utils.class_weight import compute_class_weight
from types import SimpleNamespace
_cls = np.unique(train_gen.classes)
_cw = compute_class_weight('balanced', classes=_cls, y=train_gen.classes)
class_weights = {int(c): float(w) for c, w in zip(_cls, _cw)}
print("class weights:", class_weights)

steps_per_epoch = int(np.ceil(train_gen.samples / batch_size))
loss_fn = keras.losses.CategoricalCrossentropy(label_smoothing=0.1)


def make_optimizer(peak_lr, epochs_phase, warmup_epochs=2):
    total = steps_per_epoch * epochs_phase
    warm = steps_per_epoch * warmup_epochs
    try:
        sched = keras.optimizers.schedules.CosineDecay(
            initial_learning_rate=peak_lr * 1e-2, warmup_target=peak_lr,
            warmup_steps=warm, decay_steps=max(total - warm, 1), alpha=0.02)
    except TypeError:
        sched = keras.optimizers.schedules.CosineDecay(
            initial_learning_rate=peak_lr, decay_steps=total, alpha=0.02)
    try:
        return keras.optimizers.AdamW(learning_rate=sched, weight_decay=5e-2)
    except Exception:
        return Adam(learning_rate=sched)


# ---- Phase 1: backbone frozen, train the transformer neck + head ----
phase1_epochs = 20
model.compile(optimizer=make_optimizer(3e-4, phase1_epochs), loss=loss_fn, metrics=['accuracy'])
h1 = model.fit(x=train_gen, epochs=phase1_epochs, verbose=1, validation_data=valid_gen,
               class_weight=class_weights, shuffle=False)

# ---- Phase 2: unfreeze the top of the backbone and fine-tune at a low LR ----
phase2_epochs = 20
model.unfreeze_backbone(from_layer=0.6)     # top ~40% of layers, BN kept frozen
model.compile(optimizer=make_optimizer(1e-5, phase2_epochs), loss=loss_fn, metrics=['accuracy'])
h2 = model.fit(x=train_gen, epochs=phase2_epochs, verbose=1, validation_data=valid_gen,
               class_weight=class_weights, shuffle=False)

# Merge the two phases for plotting
combined = {k: list(h1.history.get(k, [])) + list(h2.history.get(k, [])) for k in h1.history}
history = SimpleNamespace(history=combined)


# ============================================================
# Plot Training History
# ============================================================

tr_acc = history.history['accuracy']
tr_loss = history.history['loss']
val_acc = history.history['val_accuracy']
val_loss = history.history['val_loss']
index_loss = np.argmin(val_loss)
val_lowest = val_loss[index_loss]
index_acc = np.argmax(val_acc)
acc_highest = val_acc[index_acc]

Epochs = [i + 1 for i in range(len(tr_acc))]
loss_label = f'best epoch= {str(index_loss + 1)}'
acc_label = f'best epoch= {str(index_acc + 1)}'

plt.figure(figsize=(20, 8))
plt.style.use('fivethirtyeight')

plt.subplot(1, 2, 1)
plt.plot(Epochs, tr_loss, 'r', label='Training loss')
plt.plot(Epochs, val_loss, 'g', label='Validation loss')
plt.scatter(index_loss + 1, val_lowest, s=150, c='blue', label=loss_label)
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(Epochs, tr_acc, 'r', label='Training Accuracy')
plt.plot(Epochs, val_acc, 'g', label='Validation Accuracy')
plt.scatter(index_acc + 1, acc_highest, s=150, c='blue', label=acc_label)
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# Evaluate
# ============================================================

train_score = model.evaluate(train_gen, verbose=1)
valid_score = model.evaluate(valid_gen, verbose=1)
test_score = model.evaluate(test_gen, verbose=1)

print("Train Loss: ", train_score[0])
print("Train Accuracy: ", train_score[1])
print('-' * 20)
print("Validation Loss: ", valid_score[0])
print("Validation Accuracy: ", valid_score[1])
print('-' * 20)
print("Test Loss: ", test_score[0])
print("Test Accuracy: ", test_score[1])


# ============================================================
# Predictions and Confusion Matrix
# ============================================================

preds = model.predict(test_gen)
y_pred = np.argmax(preds, axis=1)

g_dict = test_gen.class_indices
classes = list(g_dict.keys())

# Confusion matrix
cm = confusion_matrix(test_gen.classes, y_pred)

plt.figure(figsize=(10, 10))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()

tick_marks = np.arange(len(classes))
plt.xticks(tick_marks, classes, rotation=45)
plt.yticks(tick_marks, classes)

thresh = cm.max() / 2.
for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
    plt.text(j, i, cm[i, j], horizontalalignment='center', color='white' if cm[i, j] > thresh else 'black')

plt.tight_layout()
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

# Classification report
print(classification_report(test_gen.classes, y_pred, target_names=classes))


# ============================================================
# Save and Reload Model  (subclassed model -> save/load weights)
# ============================================================

weights_path = '/kaggle/working/HViT_BrainTumor.weights.h5'
model.save_weights(weights_path)

loaded_model = build_model(class_count)
loaded_model.load_weights(weights_path)
loaded_model.compile(optimizer=Adam(learning_rate=1e-4),
                     loss='categorical_crossentropy', metrics=['accuracy'])


# ============================================================
# Single Image Prediction
# ============================================================

# map generator class indices -> names (correct, ordered mapping)
idx_to_class = {v: k for k, v in train_gen.class_indices.items()}

# take a test image from the test dataframe (robust; no external path needed)
image_path = test_df['filepaths'].iloc[0]
image = Image.open(image_path).convert('RGB')

img = image.resize((224, 224))
img_array = tf.keras.preprocessing.image.img_to_array(img)
img_array = tf.expand_dims(img_array, 0)          # model rescales internally

predictions = loaded_model.predict(img_array)
pred_idx = int(np.argmax(predictions))
print(f"Predicted: {idx_to_class[pred_idx]}  (true: {test_df['labels'].iloc[0]})")


# ============================================================
# Explainability: Layer-wise Relevance Propagation (LRP)
#   Transformer-adapted class-specific relevance (Chefer-style).
#   For each attention layer, relevance of a query (spatial) token i is
#       R_i = sum_j  ReLU( dS/dA_ij  *  A_ij )   averaged over heads,
#   where S is the target-class score and A the softmax attention. Each stage's
#   relevance is reshaped to its grid, upsampled to 224x224, and averaged
#   across all layers/stages to form the final relevance heatmap.
# ============================================================

def lrp_relevance(m, x, target_class=None):
    x = tf.convert_to_tensor(x, dtype=tf.float32)
    with tf.GradientTape() as tape:
        logits, attns = m(x, return_attn=True, training=False)
        if target_class is None:
            target_class = int(tf.argmax(logits[0]))
        score = logits[:, target_class]

    attn_tensors = [a for (a, H, W) in attns]
    grads = tape.gradient(score, attn_tensors)

    maps = []
    for (a, H, W), g in zip(attns, grads):
        if g is None:
            continue
        r = tf.nn.relu(g * a)                 # B, heads, Nq, Nkv
        r = tf.reduce_sum(r, axis=-1)         # B, heads, Nq   (contribution per query token)
        r = tf.reduce_mean(r, axis=1)         # B, Nq
        r = tf.reshape(r[0], [H, W])          # H, W  (batch = 1)
        r = r / (tf.reduce_max(r) + 1e-8)
        r = tf.image.resize(r[None, ..., None], (224, 224))[0, ..., 0]
        maps.append(r)

    heat = tf.reduce_mean(tf.stack(maps, axis=0), axis=0)
    heat = heat / (tf.reduce_max(heat) + 1e-8)
    return heat.numpy(), target_class


# Visualize LRP heatmaps on a few test images
n_show = 6
plt.figure(figsize=(18, 6))
for k in range(n_show):
    p = test_df['filepaths'].iloc[k]
    true_label = test_df['labels'].iloc[k]
    im = Image.open(p).convert('RGB').resize((224, 224))
    arr = tf.keras.preprocessing.image.img_to_array(im)
    arr_b = tf.expand_dims(arr, 0)

    heat, cls = lrp_relevance(loaded_model, arr_b)
    pred_label = idx_to_class[cls]

    plt.subplot(2, n_show, k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.title(f"{true_label}", fontsize=10)
    plt.axis('off')

    plt.subplot(2, n_show, n_show + k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.imshow(heat, cmap='jet', alpha=0.45)
    plt.title(f"LRP -> {pred_label}", fontsize=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import warnings
warnings.filterwarnings("ignore")

# import system libs
import os
import itertools
from PIL import Image

# import data handling tools
import numpy as np
import pandas as pd
import seaborn as sns
sns.set_style('darkgrid')
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# import Deep learning Libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print('modules loaded')


# ============================================================
# Data Preprocessing
# ============================================================

# NOTE ON THE SPLIT:
# The official Training/ and Testing/ folders of this dataset are drawn from
# DIFFERENT distributions, so training on Training/ and evaluating on Testing/
# gives a large train->test gap (models learn Training/ but fail on Testing/).
# Published 98-99% results on this dataset use k-fold CV or a re-split of the
# POOLED data. We therefore pool both folders and make a stratified
# train/val/test split so all three sets share the same distribution.

data_dirs = [
    '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Training',
    '/kaggle/input/datasets/sartajbhuvaji/brain-tumor-classification-mri/Testing',
]

filepaths = []
labels = []
for data_dir in data_dirs:
    for fold in os.listdir(data_dir):
        foldpath = os.path.join(data_dir, fold)
        if not os.path.isdir(foldpath):
            continue
        for file in os.listdir(foldpath):
            filepaths.append(os.path.join(foldpath, file))
            labels.append(fold)

# Concatenate all data paths with labels into one dataframe
Fseries = pd.Series(filepaths, name='filepaths')
Lseries = pd.Series(labels, name='labels')
all_df = pd.concat([Fseries, Lseries], axis=1)
print("total images:", len(all_df))
print(all_df['labels'].value_counts())

# Stratified split: 70% train, 15% validation, 15% test
train_df, temp_df = train_test_split(all_df, train_size=0.70, shuffle=True,
                                     stratify=all_df['labels'], random_state=42)
valid_df, test_df = train_test_split(temp_df, train_size=0.50, shuffle=True,
                                     stratify=temp_df['labels'], random_state=42)
print(f"train: {len(train_df)}  valid: {len(valid_df)}  test: {len(test_df)}")


# ============================================================
# Image Data Generators
# ============================================================

batch_size = 16
img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)

tr_gen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.06,
    height_shift_range=0.06,
    zoom_range=0.12,
    horizontal_flip=True,
    fill_mode='nearest')
ts_gen = ImageDataGenerator()

train_gen = tr_gen.flow_from_dataframe(train_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

valid_gen = ts_gen.flow_from_dataframe(valid_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                       color_mode='rgb', shuffle=True, batch_size=batch_size)

test_gen = ts_gen.flow_from_dataframe(test_df, x_col='filepaths', y_col='labels', target_size=img_size, class_mode='categorical',
                                      color_mode='rgb', shuffle=False, batch_size=batch_size)


# ============================================================
# Show sample from train data
# ============================================================

g_dict = train_gen.class_indices       # defines dictionary {'class': index}
classes = list(g_dict.keys())          # list of class names
images, labels = next(train_gen)       # get a batch of samples from the generator

plt.figure(figsize=(20, 20))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    image = images[i] / 255            # scale data to range (0 - 1)
    plt.imshow(image)
    index = np.argmax(labels[i])       # get image index
    class_name = classes[index]        # get class of image
    plt.title(class_name, color='blue', fontsize=12)
    plt.axis('off')
plt.tight_layout()
plt.show()


# ============================================================
# Model Structure
#   Pretrained-backbone Hierarchical Vision Transformer (HViT)
#     - EfficientNetV2-S (ImageNet-pretrained) hierarchical feature extractor
#     - 1x1 projection + learnable positional embedding on the 7x7 feature grid
#     - Transformer blocks (multi-head self-attention) operating on the grid,
#       each with a Dual-Branch Spatial Attention module
#       (avg-pooled channel branch + max-pooled channel branch -> spatial gate)
#     - Global average pooling + linear classification head
#     - Explainability via transformer-adapted Layer-wise Relevance Propagation
#   Requires internet ON in the Kaggle notebook for the ImageNet weights download.
# ============================================================

from tensorflow.keras.applications import EfficientNetV2S

img_size = (224, 224)
channels = 3
img_shape = (img_size[0], img_size[1], channels)
class_count = len(list(train_gen.class_indices.keys()))


DENSE_INIT = keras.initializers.TruncatedNormal(stddev=0.02)


class DropPath(layers.Layer):
    """Stochastic depth: randomly drops the residual branch per sample."""
    def __init__(self, rate=0.0, **kwargs):
        super().__init__(**kwargs)
        self.rate = rate

    def call(self, x, training=False):
        if (not training) or self.rate == 0.0:
            return x
        keep = 1.0 - self.rate
        shape = [tf.shape(x)[0]] + [1] * (len(x.shape) - 1)
        mask = tf.floor(keep + tf.random.uniform(shape, dtype=x.dtype))
        return x / keep * mask


class Attention(layers.Layer):
    """Multi-head self-attention with spatial reduction on keys/values."""
    def __init__(self, dim, num_heads, sr_ratio=1, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.sr_ratio = sr_ratio

        self.q = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        self.kv = layers.Dense(dim * 2, kernel_initializer=DENSE_INIT)
        self.proj = layers.Dense(dim, kernel_initializer=DENSE_INIT)
        if sr_ratio > 1:
            self.sr = layers.Conv2D(dim, sr_ratio, strides=sr_ratio, padding='same')
            self.sr_norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, H, W, return_attn=False):
        B = tf.shape(x)[0]
        N = tf.shape(x)[1]

        q = self.q(x)
        q = tf.reshape(q, [B, N, self.num_heads, self.head_dim])
        q = tf.transpose(q, [0, 2, 1, 3])                        # B, heads, N, hd

        if self.sr_ratio > 1:
            x_ = tf.reshape(x, [B, H, W, self.dim])
            x_ = self.sr(x_)                                     # B, H', W', C
            x_ = tf.reshape(x_, [B, -1, self.dim])
            x_ = self.sr_norm(x_)
            kv = self.kv(x_)
        else:
            kv = self.kv(x)

        Nkv = tf.shape(kv)[1]
        kv = tf.reshape(kv, [B, Nkv, 2, self.num_heads, self.head_dim])
        kv = tf.transpose(kv, [2, 0, 3, 1, 4])                   # 2, B, heads, Nkv, hd
        k, v = kv[0], kv[1]

        attn = tf.matmul(q, k, transpose_b=True) * self.scale    # B, heads, N, Nkv
        attn = tf.nn.softmax(attn, axis=-1)

        out = tf.matmul(attn, v)                                 # B, heads, N, hd
        out = tf.transpose(out, [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.dim])
        out = self.proj(out)

        if return_attn:
            return out, attn
        return out


class DualBranchSpatialAttention(layers.Layer):
    """Dual-branch spatial attention: an average-pooled channel branch and a
    max-pooled channel branch are each convolved and merged into a spatial gate."""
    def __init__(self, dim, kernel_size=7, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.branch_avg = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)
        self.branch_max = layers.Conv2D(1, kernel_size, padding='same', use_bias=False)

    def call(self, x, H, W):
        B = tf.shape(x)[0]
        f = tf.reshape(x, [B, H, W, self.dim])
        avg = tf.reduce_mean(f, axis=-1, keepdims=True)          # channel-avg descriptor
        mx = tf.reduce_max(f, axis=-1, keepdims=True)            # channel-max descriptor
        gate = tf.sigmoid(self.branch_avg(avg) + self.branch_max(mx))   # B, H, W, 1
        f = f * gate
        return tf.reshape(f, [B, H * W, self.dim])


class MLP(layers.Layer):
    def __init__(self, dim, hidden_dim, **kwargs):
        super().__init__(**kwargs)
        self.fc1 = layers.Dense(hidden_dim, kernel_initializer=DENSE_INIT)
        self.fc2 = layers.Dense(dim, kernel_initializer=DENSE_INIT)

    def call(self, x):
        x = self.fc1(x)
        x = tf.nn.gelu(x)
        x = self.fc2(x)
        return x


class Block(layers.Layer):
    """Transformer block = SR-MHSA + Dual-Branch Spatial Attention + MLP (all residual)."""
    def __init__(self, dim, num_heads, mlp_ratio=4, sr_ratio=1, drop_path=0.0, **kwargs):
        super().__init__(**kwargs)
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = Attention(dim, num_heads, sr_ratio)
        self.norm_sa = layers.LayerNormalization(epsilon=1e-6)
        self.dbsa = DualBranchSpatialAttention(dim)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = MLP(dim, dim * mlp_ratio)
        self.drop_path = DropPath(drop_path)

    def call(self, x, H, W, return_attn=False, training=False):
        if return_attn:
            y, attn = self.attn(self.norm1(x), H, W, return_attn=True)
        else:
            y = self.attn(self.norm1(x), H, W)
        x = x + self.drop_path(y, training=training)
        x = x + self.drop_path(self.dbsa(self.norm_sa(x), H, W), training=training)
        x = x + self.drop_path(self.mlp(self.norm2(x)), training=training)
        if return_attn:
            return x, attn
        return x


class HViT(Model):
    """Pretrained EfficientNetV2-S backbone -> transformer neck with Dual-Branch
    Spatial Attention -> classification head. Exposes attention maps for LRP."""
    def __init__(self, num_classes, dim=256, depth=4, num_heads=8,
                 mlp_ratio=4, drop_path=0.1, **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        # EfficientNetV2-S includes its own preprocessing, so it accepts raw 0-255 input.
        self.backbone = EfficientNetV2S(include_top=False, weights='imagenet',
                                        input_shape=(224, 224, 3), include_preprocessing=True)
        self.backbone.trainable = False                      # frozen transfer learning
        self.proj = layers.Conv2D(dim, 1, kernel_initializer=DENSE_INIT)   # 1x1: 1280 -> dim
        self.pos = self.add_weight(name='pos_embed', shape=(1, 7 * 7, dim),
                                   initializer=DENSE_INIT, trainable=True)
        dpr = list(np.linspace(0.0, drop_path, depth))
        self.blocks = [Block(dim, num_heads, mlp_ratio, sr_ratio=1, drop_path=dpr[i])
                       for i in range(depth)]
        self.norm = layers.LayerNormalization(epsilon=1e-6)
        self.drop = layers.Dropout(0.3)
        self.head = layers.Dense(num_classes, activation='softmax', kernel_initializer=DENSE_INIT)

    def call(self, x, return_attn=False, training=False):
        f = self.backbone(x, training=False)                 # B, 7, 7, 1280 (BN kept frozen)
        f = self.proj(f)                                     # B, 7, 7, dim
        H = tf.shape(f)[1]
        W = tf.shape(f)[2]
        x = tf.reshape(f, [tf.shape(f)[0], H * W, self.dim])
        x = x + self.pos
        attns = []
        for blk in self.blocks:
            if return_attn:
                x, a = blk(x, H, W, return_attn=True, training=training)
                attns.append((a, H, W))
            else:
                x = blk(x, H, W, training=training)
        x = self.norm(x)
        x = tf.reduce_mean(x, axis=1)                        # global average pooling over tokens
        x = self.drop(x, training=training)
        logits = self.head(x)
        if return_attn:
            return logits, attns
        return logits

    def unfreeze_backbone(self, from_layer=0.7):
        """Optionally unfreeze the top fraction of backbone layers for fine-tuning."""
        self.backbone.trainable = True
        cut = int(len(self.backbone.layers) * from_layer)
        for i, layer in enumerate(self.backbone.layers):
            layer.trainable = (i >= cut) and (not isinstance(layer, layers.BatchNormalization))


def build_model(num_classes):
    m = HViT(num_classes)
    _ = m(tf.zeros((1, img_size[0], img_size[1], channels)))   # build the graph
    return m


model = build_model(class_count)
model.summary()


# ============================================================
# Training
# ============================================================

# Balance the classes (no_tumor is under-represented).
from sklearn.utils.class_weight import compute_class_weight
from types import SimpleNamespace
_cls = np.unique(train_gen.classes)
_cw = compute_class_weight('balanced', classes=_cls, y=train_gen.classes)
class_weights = {int(c): float(w) for c, w in zip(_cls, _cw)}
print("class weights:", class_weights)

steps_per_epoch = int(np.ceil(train_gen.samples / batch_size))
loss_fn = keras.losses.CategoricalCrossentropy(label_smoothing=0.1)


def make_optimizer(peak_lr, epochs_phase, warmup_epochs=2):
    total = steps_per_epoch * epochs_phase
    warm = steps_per_epoch * warmup_epochs
    try:
        sched = keras.optimizers.schedules.CosineDecay(
            initial_learning_rate=peak_lr * 1e-2, warmup_target=peak_lr,
            warmup_steps=warm, decay_steps=max(total - warm, 1), alpha=0.02)
    except TypeError:
        sched = keras.optimizers.schedules.CosineDecay(
            initial_learning_rate=peak_lr, decay_steps=total, alpha=0.02)
    try:
        return keras.optimizers.AdamW(learning_rate=sched, weight_decay=5e-2)
    except Exception:
        return Adam(learning_rate=sched)


# ---- Phase 1: backbone frozen, train the transformer neck + head ----
phase1_epochs = 30
model.compile(optimizer=make_optimizer(3e-4, phase1_epochs), loss=loss_fn, metrics=['accuracy'])
h1 = model.fit(x=train_gen, epochs=phase1_epochs, verbose=1, validation_data=valid_gen,
               class_weight=class_weights, shuffle=False)

# ---- Phase 2: unfreeze the top of the backbone and fine-tune at a low LR ----
phase2_epochs = 20
model.unfreeze_backbone(from_layer=0.6)     # top ~40% of layers, BN kept frozen
model.compile(optimizer=make_optimizer(1e-5, phase2_epochs), loss=loss_fn, metrics=['accuracy'])
h2 = model.fit(x=train_gen, epochs=phase2_epochs, verbose=1, validation_data=valid_gen,
               class_weight=class_weights, shuffle=False)

# Merge the two phases for plotting
combined = {k: list(h1.history.get(k, [])) + list(h2.history.get(k, [])) for k in h1.history}
history = SimpleNamespace(history=combined)


# ============================================================
# Plot Training History
# ============================================================

tr_acc = history.history['accuracy']
tr_loss = history.history['loss']
val_acc = history.history['val_accuracy']
val_loss = history.history['val_loss']
index_loss = np.argmin(val_loss)
val_lowest = val_loss[index_loss]
index_acc = np.argmax(val_acc)
acc_highest = val_acc[index_acc]

Epochs = [i + 1 for i in range(len(tr_acc))]
loss_label = f'best epoch= {str(index_loss + 1)}'
acc_label = f'best epoch= {str(index_acc + 1)}'

plt.figure(figsize=(20, 8))
plt.style.use('fivethirtyeight')

plt.subplot(1, 2, 1)
plt.plot(Epochs, tr_loss, 'r', label='Training loss')
plt.plot(Epochs, val_loss, 'g', label='Validation loss')
plt.scatter(index_loss + 1, val_lowest, s=150, c='blue', label=loss_label)
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(Epochs, tr_acc, 'r', label='Training Accuracy')
plt.plot(Epochs, val_acc, 'g', label='Validation Accuracy')
plt.scatter(index_acc + 1, acc_highest, s=150, c='blue', label=acc_label)
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# Evaluate
# ============================================================

train_score = model.evaluate(train_gen, verbose=1)
valid_score = model.evaluate(valid_gen, verbose=1)
test_score = model.evaluate(test_gen, verbose=1)

print("Train Loss: ", train_score[0])
print("Train Accuracy: ", train_score[1])
print('-' * 20)
print("Validation Loss: ", valid_score[0])
print("Validation Accuracy: ", valid_score[1])
print('-' * 20)
print("Test Loss: ", test_score[0])
print("Test Accuracy: ", test_score[1])


# ============================================================
# Predictions and Confusion Matrix
# ============================================================

preds = model.predict(test_gen)
y_pred = np.argmax(preds, axis=1)

g_dict = test_gen.class_indices
classes = list(g_dict.keys())

# Confusion matrix
cm = confusion_matrix(test_gen.classes, y_pred)

plt.figure(figsize=(10, 10))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()

tick_marks = np.arange(len(classes))
plt.xticks(tick_marks, classes, rotation=45)
plt.yticks(tick_marks, classes)

thresh = cm.max() / 2.
for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
    plt.text(j, i, cm[i, j], horizontalalignment='center', color='white' if cm[i, j] > thresh else 'black')

plt.tight_layout()
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

# Classification report
print(classification_report(test_gen.classes, y_pred, target_names=classes))


# ============================================================
# Save and Reload Model  (subclassed model -> save/load weights)
# ============================================================

weights_path = '/kaggle/working/HViT_BrainTumor.weights.h5'
model.save_weights(weights_path)

loaded_model = build_model(class_count)
loaded_model.load_weights(weights_path)
loaded_model.compile(optimizer=Adam(learning_rate=1e-4),
                     loss='categorical_crossentropy', metrics=['accuracy'])


# ============================================================
# Single Image Prediction
# ============================================================

# map generator class indices -> names (correct, ordered mapping)
idx_to_class = {v: k for k, v in train_gen.class_indices.items()}

# take a test image from the test dataframe (robust; no external path needed)
image_path = test_df['filepaths'].iloc[0]
image = Image.open(image_path).convert('RGB')

img = image.resize((224, 224))
img_array = tf.keras.preprocessing.image.img_to_array(img)
img_array = tf.expand_dims(img_array, 0)          # model rescales internally

predictions = loaded_model.predict(img_array)
pred_idx = int(np.argmax(predictions))
print(f"Predicted: {idx_to_class[pred_idx]}  (true: {test_df['labels'].iloc[0]})")


# ============================================================
# Explainability: Layer-wise Relevance Propagation (LRP)
#   Transformer-adapted class-specific relevance (Chefer-style).
#   For each attention layer, relevance of a query (spatial) token i is
#       R_i = sum_j  ReLU( dS/dA_ij  *  A_ij )   averaged over heads,
#   where S is the target-class score and A the softmax attention. Each stage's
#   relevance is reshaped to its grid, upsampled to 224x224, and averaged
#   across all layers/stages to form the final relevance heatmap.
# ============================================================

def lrp_relevance(m, x, target_class=None):
    x = tf.convert_to_tensor(x, dtype=tf.float32)
    with tf.GradientTape() as tape:
        logits, attns = m(x, return_attn=True, training=False)
        if target_class is None:
            target_class = int(tf.argmax(logits[0]))
        score = logits[:, target_class]

    attn_tensors = [a for (a, H, W) in attns]
    grads = tape.gradient(score, attn_tensors)

    maps = []
    for (a, H, W), g in zip(attns, grads):
        if g is None:
            continue
        r = tf.nn.relu(g * a)                 # B, heads, Nq, Nkv
        r = tf.reduce_sum(r, axis=-1)         # B, heads, Nq   (contribution per query token)
        r = tf.reduce_mean(r, axis=1)         # B, Nq
        r = tf.reshape(r[0], [H, W])          # H, W  (batch = 1)
        r = r / (tf.reduce_max(r) + 1e-8)
        r = tf.image.resize(r[None, ..., None], (224, 224))[0, ..., 0]
        maps.append(r)

    heat = tf.reduce_mean(tf.stack(maps, axis=0), axis=0)
    heat = heat / (tf.reduce_max(heat) + 1e-8)
    return heat.numpy(), target_class


# Visualize LRP heatmaps on a few test images
n_show = 6
plt.figure(figsize=(18, 6))
for k in range(n_show):
    p = test_df['filepaths'].iloc[k]
    true_label = test_df['labels'].iloc[k]
    im = Image.open(p).convert('RGB').resize((224, 224))
    arr = tf.keras.preprocessing.image.img_to_array(im)
    arr_b = tf.expand_dims(arr, 0)

    heat, cls = lrp_relevance(loaded_model, arr_b)
    pred_label = idx_to_class[cls]

    plt.subplot(2, n_show, k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.title(f"{true_label}", fontsize=10)
    plt.axis('off')

    plt.subplot(2, n_show, n_show + k + 1)
    plt.imshow(np.array(im).astype('uint8'))
    plt.imshow(heat, cmap='jet', alpha=0.45)
    plt.title(f"LRP -> {pred_label}", fontsize=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
